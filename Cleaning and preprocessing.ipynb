# VALIDATING ROI COLUMN

```python
df.dropna(subset=['Revenue','Acquisition_Cost','Conversions'],inplace=True)
```

```python
valid_roi = (df['Revenue'] - df['Conversions'] * df['Acquisition_Cost']) / (df['Conversions'] * df['Acquisition_Cost'])
invalid_roi = df['ROI'].isnull() | (df['ROI'] < -1) | ((df['ROI'] - valid_roi).abs() > 0.01)

#  Replacing invalid ROI values with calculated ones
df['ROI'] = np.where(invalid_roi, valid_roi, df['ROI'])
```

# DATA CLEANING

```python
df.info()
```

```python
# Dropping Campaign_ID and Date columns as they don't add value

df.drop(columns = ['Campaign_ID','Date'],inplace = True)
```

```python
# Imputing categorical columns with the value "Unknown"

cat_cols = ['Campaign_Type','Target_Audience','Language','Customer_Segment','Channel_Used']
df[cat_cols] = df[cat_cols].fillna("Unknown")
```

```python
# Calculating skewness in numerical columns for null imputation

numericals = ['Duration','Impressions','Clicks','Leads','Conversions','Engagement_Score']
df[numericals].skew()
```

```python
for i in numericals:
    skew_value = df[i].skew()
    if abs(skew_value) > 0.5:
        df[i] = df[i].fillna(df[i].median())
    else:
        df[i] = df[i].fillna(df[i].mean())
```

```python
# Creating a new feature - Profit / Loss flag

df['Profit/Loss_flag'] = np.where(df['ROI'] > 0, 1, 0)
```

```python
# Standardizing datatypes

int_cols = ['Duration','Impressions','Clicks','Leads','Conversions','Profit/Loss_flag']
float_cols = ['Revenue','Acquisition_Cost','ROI','Engagement_Score']

for i in int_cols:
    df[i] = df[i].astype('int')

for i in float_cols:
    df[i] = df[i].astype('float')
```

```python
df.info()
```

# OUTLIER HANDLING

```python
num_cols = ['Duration','Impressions','Clicks','Leads','Conversions',
            'Acquisition_Cost','Engagement_Score','Revenue','ROI']

# calculating the percentage of outliers in each column

for col in num_cols:
    Q1, Q3 = df[col].quantile([0.25, 0.75])
    IQR = Q3 - Q1
    n = ((df[col] < Q1 - 1.5 * IQR) | (df[col] > Q3 + 1.5 * IQR)).sum()
    print(f'{col:<18} outliers: {n:>6} ({n / len(df) * 100:.1f}%)')
```

```python

for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    # metrics cannot be negative numbers
    if lower_limit < 0 and col != "ROI":
        lower_limit = 0

    df[col] = df[col].clip(lower=lower_limit, upper=upper_limit)
    print(f"Column: {col:<18} | Cleaned Limits: [{lower_limit:.2f}, {upper_limit:.2f}]")
```

```python
fig, axes = plt.subplots(3, 3, figsize=(15, 10))

for ax, col in zip(axes.ravel(), df.select_dtypes(include="number").columns):
    sns.boxplot(y=df[col], ax=ax, color='lightsteelblue')
    ax.set_title(col)

plt.tight_layout()
plt.show()
```
